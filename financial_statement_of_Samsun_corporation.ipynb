```python
import requests
import pandas as pd
from IPython.display import display

api_key = 'personnal_API_key'
corp_code = '00126380'
bsns_year = '2025'
reprt_code = '11011'

url = 'https://opendart.fss.or.kr/api/fnlttSinglAcnt.json'
params = {
    'crtfc_key': api_key,
    'corp_code': corp_code,
    'bsns_year': bsns_year,
    'reprt_code': reprt_code
}

response = requests.get(url, params = params)
data = response.json()

if data['status'] == '000':
    df = pd.DataFrame(data['list'])

    bs_df = df[df['sj_div'] == 'BS']

    result_df = bs_df[['account_nm', 'thstrm_amount']]
    print('extracting financial statement of Samsung corporation is complete')
    display(result_df)
else:
    print("request failed", data['message'])
```
